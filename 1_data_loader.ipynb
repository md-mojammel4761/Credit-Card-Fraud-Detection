```python
value = df['is_fraud'].value_counts()

plt.figure(figsize=(10,8))
plt.pie(
    value,
    labels=value.index,
    startangle=90,
    explode=[0.05] * len(value),
    autopct='%1.1f%%',
    shadow=True,
    colors= plt.cm.Set2.colors
)
plt.title('IS_Fraud Distribution',fontsize=16,weight='bold')
plt.legend(title='IS_Fraud', labels=['NO  (0)','YES (1)'])
plt.show()
```
