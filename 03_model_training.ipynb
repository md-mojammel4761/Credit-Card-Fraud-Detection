```python
import joblib
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
xtrain = joblib.load('models/xtrain.jbl')
xtest = joblib.load('models/xtest.jbl')
ytrain = joblib.load('models/ytrain.jbl')
ytest = joblib.load('models/ytest.jbl')
```

```python
models = {
    'Logistic_Regression':LogisticRegression(),
    'RandomForest_Classifier':RandomForestClassifier(),
    'XGB_Classifier':XGBClassifier()
}
```

```python
trained_models = {}

for name, model in models.items():

    model.fit(xtrain, ytrain)

    trained_models[name] = model

    print(f'{name} trained successfully')
```

```python
joblib.dump(trained_models['Logistic_Regression'],'models/log_reg.jbl')
joblib.dump(trained_models['XGB_Classifier'],'models/xgb_clf.jbl')
joblib.dump(trained_models['RandomForest_Classifier'],'models/rf.jbl')
print('All models saved successfully')
```
